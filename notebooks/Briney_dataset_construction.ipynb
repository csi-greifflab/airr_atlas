```python
import pandas as pd
from collections import Counter
```

#### CONSTRUCTING THE ORIGINAL DATASET ####

```python
dirs_for_sampling = ['316188','326650','326651','326713','326737','326780','326797','326907','327059','D103']
replicate_files = ['1_consensus.txt','2_consensus.txt','3_consensus.txt','4_consensus.txt','5_consensus.txt','6_consensus.txt','7_consensus.txt','8_consensus.txt','9_consensus.txt','10_consensus.txt','11_consensus.txt','12_consensus.txt','13_consensus.txt','14_consensus.txt','15_consensus.txt','16_consensus.txt','17_consensus.txt','18_consensus.txt']
replicates = []
combinations = [(x,y) for x in dirs_for_sampling for y in replicate_files]
for com in combinations:
	replicates.append('./'+com[0]+'/'+com[1])
li = []
for filename in replicates:
	batch_name = filename.split('/')[-2]+'_'+filename.split('/')[-1]
	print(batch_name)
	tmp_df = pd.read_csv(filename, index_col=None, header=0)
	tmp_df = tmp_df[tmp_df['chain'] == 'heavy']
	tmp_df = tmp_df.dropna(subset=['cdr3_aa'])
	tmp_df = tmp_df[tmp_df['productive'] == 'yes']
	tmp_df['batch'] = batch_name
	if len(tmp_df.index.tolist())>100000:
		tmp_df = tmp_df.sample(n=100000)
	print('Adding',tmp_df.shape)
	li.append(tmp_df)
united_dataset = pd.concat(li,ignore_index=True,axis=0)
united_dataset['index'] = united_dataset.index + 1
print(Counter(united_dataset['batch'].tolist()))
united_dataset.to_csv('briney_original_dataset.csv',index=None)
to_fasta = united_dataset['cdr3_aa'].tolist()
with open('briney_original_dataset.fasta','w') as f_o:
	for i in range(len(to_fasta)):
		f_o.write('>'+str(i)+'\n')
		f_o.write(to_fasta[i]+'\n')
```

#### CONSTRUCTING THE FLANKED DATASET ####

```python
dirs_for_sampling = ['316188','326650','326651','326713','326737','326780','326797','326907','327059','D103']
replicate_files = ['1_consensus.txt','2_consensus.txt','3_consensus.txt','4_consensus.txt','5_consensus.txt','6_consensus.txt','7_consensus.txt','8_consensus.txt','9_consensus.txt','10_consensus.txt','11_consensus.txt','12_consensus.txt','13_consensus.txt','14_consensus.txt','15_consensus.txt','16_consensus.txt','17_consensus.txt','18_consensus.txt']
replicates = []
combinations = [(x,y) for x in dirs_for_sampling for y in replicate_files]
for com in combinations:
	replicates.append('./'+com[0]+'/'+com[1])
li = []
for filename in replicates:
	batch_name = filename.split('/')[-2]+'_'+filename.split('/')[-1]
	print(batch_name)
	tmp_df = pd.read_csv(filename, index_col=None, header=0)
	tmp_df = tmp_df.dropna(subset=['cdr3_aa'])
	tmp_df['cdr3_aa'] = tmp_df['cdr3_aa'].apply(lambda x: 'C' + x if not x.startswith('C') else x)
	tmp_df['cdr3_aa'] = tmp_df['cdr3_aa'].apply(lambda x: x + 'W' if not x.endswith('W') else x)
	tmp_df = tmp_df[tmp_df['chain'] == 'heavy']
	tmp_df = tmp_df[tmp_df['productive'] == 'yes']
	tmp_df['batch'] = batch_name
	if len(tmp_df.index.tolist())>100000:
		tmp_df = tmp_df.sample(n=100000)
	print('Adding',tmp_df.shape)
	li.append(tmp_df)
united_dataset = pd.concat(li,ignore_index=True,axis=0)
united_dataset['index'] = united_dataset.index + 1
print(Counter(united_dataset['batch'].tolist()))
united_dataset.to_csv('briney_only_flanked.csv',index=None)
to_fasta = united_dataset['cdr3_aa'].tolist()
with open('briney_only_flanked.fasta','w') as f_o:
	for i in range(len(to_fasta)):
		f_o.write('>'+str(i)+'\n')
		f_o.write(to_fasta[i]+'\n')
```
